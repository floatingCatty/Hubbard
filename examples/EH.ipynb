```python
from hubbard.solver import ED_solver, DMRG_solver, NQS_solver
import numpy as np

# H = e_0 c_0^* c_0 + t01 c_0^*c_1 + c.c. + e_1
# solver = DMRG_solver(
#     n_int=18, # interacting sites
#     n_noint=0, # no-interacting sites (bath), 0 if not impurity system
#     n_elec=9, # number of electrons
#     nspin=4, # 1: spin-degenerate, 2: spin collinear, 4: spin noncollinear
#     nupdate=8,
#     n_threads=40,

#     # optional
#     bond_dim=4000,
#     reorder=True,
#     iprint=1
# )

# solver = ED_solver(
#     n_int=8, # interacting sites
#     n_noint=0, # no-interacting sites (bath), 0 if not impurity system
#     n_particle=[(0,4),(1,3),(2,2),(3,1),(4,0)],
#     nspin=4,
# )

solver = NQS_solver(
        n_int=8, 
        n_noint=0,
        n_particle=(2,2),
        nspin=1,
        mfepmax=50,
        nnepmax=50,
        d_emb=4,
        nblocks=3,
        hidden_channels=8,
        out_channels=8,
        ffn_hidden=[8],
        heads=4,
        Ptol=1e-4, # this value should be smaller than at least 2e-4 if expecting 1e-3 convergence
        Etol=6e-4,
        debug_neural=True,
    )
```

```python
solver.solve(T=np.zeros((16,16)), intparam={"int_type": "EH", "chem_1":0, "chem_2":0, "Nx":2, "Ny":2})
```

```python
import matplotlib.pyplot as plt
RDM = solver.cal_RDM(solver.vec) # ED
# RDM = solver.RDM() # DMRG
# RDM = solver.Psample(solver.state, RDM=True)
```

```python
plt.matshow(RDM, vmax=1.0, vmin=-1.0, cmap="bwr")
plt.colorbar()
plt.show()
```
